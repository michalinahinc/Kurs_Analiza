```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as ss
%matplotlib inline 
%config InlineBackend.figure_format = 'retina'

katalog = ".\\data\\"  

df = pd.read_csv(katalog + "apartments_for_rent_classified_100K.csv", sep=';', encoding="cp1250")
# df = pd.read_csv(katalog + "apartments_for_rent_classified_3000.csv", sep=';', encoding="cp1250")
# df.head(2)

df = df.dropna(subset=["cityname"])
df = df.dropna(subset=["price"])
df["pets_allowed"] = df["pets_allowed"].fillna("None")
df.isna().sum()  
```


Sprawdziłam wyrywkowo wiersze, dla których brakuje danych o łazience. Prawdopodobnie te łazienki są, ale z analizy danych ofert, które oglądałam szukając ich adresów wynika, że nie jest to 100% pewne. Trafiłam na dokładnie te oferty, które są w pliku, a także inne, z tych samych budynków. 
Po wcześniejszym usunięciu złych danych geograficznych i linii "bezcennych" oraz zmianie dotyczącej zwierzaków pozostały 59 dwie linie bez łazienek, które proponuję usunąć.

## <span style="color: green;">WNIOSEK - pomijamy 59 linie bez wskazanej ilości łazienek</span>

```python
df = df.dropna(subset=["bathrooms"])
df.isna().sum()  
```

Sprawdziłam wyrywkowo linie, dla których przypisano 0 sypialni. Są to :
- nieduże pokoje <span style="color: green;">square_feet < 700</span> łączące wiele funkcji typu:</BR>
    <span style="color: green;">kącik kuchenny + sypialny + TV</span></BR>
    Te dane bym przyjęła jako poprawne
- większe pomieszczenia <span style="color: orange;">square_feet >= 700</span>, dla których błędnie nie przypasano sypialni, mimo, że w ofercie obecnie je widać </BR>
    Takich linii jest 90. Te dane bym usunęła

Podobną zasadę 700 stóp można użyć dla 125 linii, które mają nieokreśloną ilość sypialni.    

```python
# znajdź indeksy wierszy do usunięcia
idx_to_drop = df[((df["bedrooms"] == 0) | (df["bedrooms"].isna()))& (df["square_feet"] >= 700)].index

# usuń te wiersze
df = df.drop(idx_to_drop)
# df.to_csv("plik_wynik.csv", index=False, sep=";")   # to dla sprawdzenia jak poszło ....  ;)
df["bedrooms"] = df["bedrooms"].fillna(0)
df.isna().sum() 
```

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as ss
%matplotlib inline 
%config InlineBackend.figure_format = 'retina'

df = pd.read_csv("../data/apartments_for_rent_classified_100K.csv", sep=';', encoding="cp1250")
# df.head(2)

df.isna().sum()  
```


Sprawdziłam wyrywkowo wiersze, dla których brakuje danych o łazience. Prawdopodobnie te łazienki są, ale z analizy danych ofert, które oglądałam szukając ich adresów wynika, że nie jest to 100% pewne. Trafiłam w sieci na dokładnie te oferty, które są w pliku, a także inne, z tych samych budynków. 
Wierszy dla których brakuje danych o łazience jest ich stosunkowo niewiele (mniej niż 60, jeżeli wcześniej wyczyścimy inne dane). <span style="color: green;">Linie bez łazienek, które proponuję usunąć</span>.


```python
df = df.dropna(subset=["bathrooms"])
df.isna().sum()  
```

Sprawdziłam wyrywkowo w sieci linie, dla których przypisano 0 sypialni. </BR>
Wzięłam pod uwagę, ze 500 ft² , to 46,45 m² h.</BR> Możemy dyskutować, czy przypadkiem nie zmienić tej liczby, bo w ofertach widziałam pomieszczenia bez sypialni mające 700 ft². Jednak chyba 500 ft² jest liczbą bezpieczniejszą.</BR>
Pomieszczenia te podzieliłabym na:
- nieduże pomieszczenia <span style="color: green;">square_feet < 500</span> łączące wiele funkcji typu:</BR>
    <span style="color: green;">kącik kuchenny + sypialny + TV</span></BR>
    Te dane bym przyjęła jako poprawne
- większe pomieszczenia <span style="color: orange;">square_feet >= 500</span>, dla których błędnie nie przypasano sypialni, mimo, że w ofercie obecnie je widać </BR>
    Takich linii jest 90. Te dane bym usunęła

Podobną zasadę 500 ft² można użyć dla 125 linii, które mają nieokreśloną ilość sypialni.    

```python
# znajdź indeksy wierszy do usunięcia dla ilości = 0 i =null
idx_to_drop = df[((df["bedrooms"] == 0) | (df["bedrooms"].isna()))& (df["square_feet"] >= 500)].index

# usuń te wiersze
df = df.drop(idx_to_drop)
# df.to_csv("plik_wynik.csv", index=False, sep=";")   # to dla sprawdzenia jak poszło ....  ;)

# dla małych pomieszczeń, które pozostały z wartością null zmień ja na = 0
df["bedrooms"] = df["bedrooms"].fillna(0)
df.isna().sum() 
```
